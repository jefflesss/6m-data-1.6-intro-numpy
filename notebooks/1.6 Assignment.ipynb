## 3\. Boolean masks: Work with targets

```python
names = np.array(["Ana", "Ben", "Chen", "Dana", "Eli", "Fatima", "George", "Hui"])
spend = np.array([200, 150, 300, 120, 180, 220, 160, 310])      # marketing spend
revenue = np.array([400, 180, 500, 100, 220, 260, 150, 600])   # revenue
```

```python
#1. Compute the **ROI** for each person: `roi = revenue / spend`.
roi = revenue / spend
print("\n ROI:", roi)
#2. Create a boolean mask for customers with `roi >= 2.0`.  
mask_roi = (roi >= 2.0)
#3. Use the mask to:  List their names and List their spend and revenue 
print("\nNames with ROI >= 2.0\n", names[mask_roi])
print("\nSpend with ROI >= 2.0\n",spend[mask_roi])
print("\nRevenue ROI >= 2.0\n", revenue[mask_roi])

#4. Create a second mask for customers with `spend >= 200`.
spend_mask = (spend >= 200)
print("\nNames spend >= 200\n", names[spend_mask])

#5. Combine the two masks to find customers who have **roi \>= 2.0 AND spend \>= 200**.
combined_mask = mask_roi & spend_mask
print("\nCustomers w ROI >= 2 and spend >= 200\n", names[combined_mask])
```

## 4\. Gradebook: Aggregations and broadcasting

```python
students = np.array(["S1", "S2", "S3", "S4", "S5"])
subjects = np.array(["Math", "Stats", "Python"])

scores = np.array([
    [75, 80, 85],
    [60, 65, 70],
    [90, 88, 92],
    [82, 79, 84],
    [70, 72, 78]
])
```

```python
#1. Compute each student’s **average score** (per row).
avg_student_score = np.mean(scores, axis=1)
print("\nAverage score per student:\n", avg_student_score.reshape(5,1))
#2. Compute each subject’s **average score** (per column).  
avg_subject_score = np.mean(scores, axis = 0)
print("\nAverage score per subject:\n", avg_subject_score)
#3. Create `scores_centered` by subtracting the **subject mean** from each column (broadcasting).  
scores_centered = scores - avg_subject_score
print("\nScores before\n",scores)
print("\nScores centered\n",scores_centered)
```

```python
#4. For `scores_centered`, compute per-student averages again.  
scores_centered_mean = np.mean(scores_centered, axis=1)
print("\nAverage score per student:\n", avg_student_score.reshape(5,1))
print("\nScores centered mean\n",scores_centered_mean.reshape(5,1))
```

## **Part 5: Reshape & Flatten**

```python
#**Scenario:** You have 30 days of daily website visits, and you want to summarize them by week.

daily_visits = np.array([
    120, 130, 125, 140, 150,
    160, 170, 155, 145, 135,
    128, 132, 138, 142, 148,
    152, 158, 162, 168, 172,
    180, 190, 185, 175, 165,
    155, 145, 135, 125, 115
])
```

```python
#1. Reshape `daily_visits` into a `(6, 5)` array representing **6 weeks × 5 days**. 
reshape_daily_visits = daily_visits.reshape(6,5)
print("\nReshaped daily visits, 6 weeks 5 days\n", reshape_daily_visits) 
#2. Compute **total visits per week**.
total_visits_weekly = np.sum(reshape_daily_visits, axis=1)
print("\nTotal visits per week\n",total_visits_weekly.reshape(6,1))
##3. Compute **average visits per day of the week** (e.g., all “day 1 of week” together, all “day 2 of week” together, etc.).  
avg_per_day = np.mean(reshape_daily_visits, axis = 0)
print("\n Average per day\n", avg_per_day)
##4. Flatten the reshaped array back to 1D and confirm it matches the original `daily_visits`.
flatten = reshape_daily_visits.reshape(1,30)
print("\n Flattened reshaped array\n", flatten)
print("\n Original array flattened \n", daily_visits.reshape(1,30))
```
